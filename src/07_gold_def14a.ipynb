### Creating gold fact table

The Gold fact table contains year‑level executive compensation data for each company (CIK).
It is built from the Silver DEF 14A table and provides a clean, analytics‑ready structure for Power B

```python
from pyspark.sql import functions as F

silver = spark.table("sec_def14a_silver")

gold_exec_comp = (
    silver.select(
        "cik",
        "fiscal_year",
        "ticker",
        "company_name",
        "form",
        "peo_name",
        "peo_total_comp",
        "peo_actually_paid_comp",
        "neo_avg_total_comp",
        "neo_avg_actually_paid_comp",
        "net_income",
        "total_shareholder_return",
        "peer_group_tsr",
        "company_selected_measure",
        "company_selected_measure_value",
        "is_latest_peo",
        "is_latest_neo",
        "is_latest_net_income",
        "ingestion_ts"
    )
)

gold_exec_comp.write.format("delta").mode("overwrite").saveAsTable("gold_sec_def14a")

print("Gold Exec Compensation Fact created.")
```
